# Introduction
This analysis is to use the supply chain dataset to find out any possible problems with suppliers or any shortage with an inventory.
Questions to help analysis:
* how many suppliers
* how many number of types of products are supplied by each supplier
* how many products have failed  in inspection, and who is the supplier
* how many products are below their stock levels
* Is a one-time order enough for those products to reach their stock level


```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('supply_chain_data.csv')
```

```python
supply = df.copy()
```

```python
pd.set_option('display.max_columns', None)
supply.head(5)
```

```python
supply.info()
```

```python
supply.shape
```

**How many Supplier?**

```python
unique_suppliers = supply['Supplier name'].nunique()
print(f"Number of unique suppliers: {unique_suppliers}")
print("Suppliers:", supply['Supplier name'].unique())
```

**how many number of types of products are supplied by each supplier**

```python
supply.groupby('Supplier name')['Product type'].value_counts()
```


**how many products have failed  in inspection, and who is the supplier**

```python
failed_Inspections = supply[supply['Inspection results'] =='Fail']
failed_Inspections.groupby('Supplier name')['Inspection results'].value_counts()
```

**how many products are below their stock levels**

```python
below_stocks = supply[supply['Stock levels']-supply['Availability'] >= 0]
below_stocks.head(5)
```

 **Is a one-time order enough for those products to reach their stock level?**

```python
below_stocks['new_avail'] = below_stocks['Availability']+below_stocks['Order quantities']
```

```python
below_stocks['enough stocks'] = below_stocks['new_avail'] >= below_stocks['Stock levels']
```

```python
below_stocks[below_stocks['enough stocks']==False]
```
